```python
import re
import matplotlib.pyplot as plt

# Path to your log file
log_file = "training_logs_biharmonic.txt"

# Read the file and extract epoch and loss values
epochs = []
losses = []

with open(log_file, "r") as f:
    for line in f:
        match = re.search(r"\[Epoch (\d+)/\d+\] loss = ([0-9\.eE+-]+)", line)
        if match:
            epoch = int(match.group(1))
            loss = float(match.group(2))
            epochs.append(epoch)
            losses.append(loss)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(epochs, losses, marker="o")
plt.title("Training Loss over Epochs")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.yscale("log")
plt.grid(True)
plt.show()
```
